# Bronze Layer

## Libraries

```python
from pyspark.sql.functions import when, col
```

## Load data

```python
# Read raw data
raw_data = spark.table('workspace.telco.raw_data')
display(raw_data.limit(5)) # Sample
```

## Fix data types

```python
# Convert TotalCharges from string to double
clean_data = raw_data.withColumn(
    "TotalCharges",
    when(col("TotalCharges") == " ", None)
    .otherwise(col("TotalCharges").cast("double"))
)
```

## Handling missing values

```python
# Count missing values
missing_counts = [(c, clean_data.filter(col(c).isNull()).count()) for c in clean_data.columns]
display(missing_counts)
```

```python
# Drop missing values
clean_data = clean_data.dropna()
```

## Saving Bronze Data

```python
# Saving bronze layer data
clean_data.write.mode("overwrite").saveAsTable("workspace.telco.bronze_data")
```
